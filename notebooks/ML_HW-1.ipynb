Выполните задание 11 с помощью циклов. Сравните время выполнения

```python
%%time
# с помощью списковых включений
[data[data['race/ethnicity']==group]['reading score'].mean() for group in data['race/ethnicity'].unique()]
```

```python
%%time
# с помощью списковых включений
[data[data['education']==edu]['writing score'].min() for edu in data['education'].unique()]
```

```python
%%time
# с помощью обычного цикла
for group in data['race/ethnicity'].unique():
    print(data[data['race/ethnicity']==group]['reading score'].mean())
```

```python
%%time
# # с помощью обычного цикла
for edu in data['education'].unique():
    print(data[data['education']==edu]['writing score'].min())
```

```python
%%time
# с помощью прохождения по всем строкам датасета
race_data = {}
edu_min = {}

for i in range(len(data)):
    race = data['race/ethnicity'].iloc[i]
    reading = data['reading score'].iloc[i]
    edu = data['education'].iloc[i]
    writing = data['writing score'].iloc[i]

    if race not in race_data:
        race_data[race] = [0, 0] # sum, count
    race_data[race][0] += reading
    race_data[race][1] += 1

    if edu not in edu_min or writing < edu_min[edu]:
        edu_min[edu] = writing

print('Средний бал по чтению')
for race in sorted(race_data.keys()):
    avg = race_data[race][0]/race_data[race][1]
    print(f'{race} - {avg:.2f}')
print()

print('Минимальный бал по письму')
for edu in sorted(edu_min.keys()):
    print(f'{edu} - {edu_min[edu]}')
```

Задачу предсказания оценки можно решать как регрессией, так и классификацией. Что нужно сделать для того, чтобы решать задачу с помощью классификации, если нам важна оценка с точностью не до балла, а условный диапазон, например, отлично, хорошо, удовлетворительно и так далее? Проведите необходимые преобразования над числовыми колонками.

**Ваш ответ здесь...**

Нужно создать новую целевую переменную (новую колонку) ввиде категориальных данных на основании числовых данных из соответствующей колонки.
Например, если нас интересует предсказание оценки по математике, то преобразовываем ее данные разбив по диапазонам и присвоим им соответствующие категорияльные значения "неудовлетворительно", "удовлетворительно", "хорошо", "отлично". Если нужны предсказания и по другим предметам, то проделываем те же манипуляции и с другими колонками.

```python
# ваш код здесь
def mark_degrees(x):
    mark_degree = ''
    if x < 50:
        mark_degree = 'неудовлетворительно'
    elif x < 70:
        mark_degree = 'удовлетворительно'
    elif x < 90:
        mark_degree = 'хорошо'
    else:
        mark_degree = 'отлично'
    return mark_degree    
```
